## Data Cleaning & Handling

```python
df.isna().sum()
```

```python
# dropping unnecessary columns
df = df.drop(columns=['board', 'room'])
```

```python
# filling numerical columns with median
df['# appli. rec\'d'] = df['# appli. rec\'d'].fillna(df['# appli. rec\'d'].median())
df['# appl. accepted'] = df['# appl. accepted'].fillna(df['# appl. accepted'].median())
df['# new stud. enrolled'] = df['# new stud. enrolled'].fillna(df['# new stud. enrolled'].median())
df['# FT undergrad'] = df['# FT undergrad'].fillna(df['# FT undergrad'].median())
df['# PT undergrad'] = df['# PT undergrad'].fillna(df['# PT undergrad'].median())
df['in-state tuition'] = df['in-state tuition'].fillna(df['in-state tuition'].median())
df['out-of-state tuition'] = df['out-of-state tuition'].fillna(df['out-of-state tuition'].median())
df['estim. book costs'] = df['estim. book costs'].fillna(df['estim. book costs'].median())
df['% fac. w/PHD'] = df['% fac. w/PHD'].fillna(df['% fac. w/PHD'].median())
df['stud./fac. ratio'] = df['stud./fac. ratio'].fillna(df['stud./fac. ratio'].median())
df['Graduation rate'] = df['Graduation rate'].fillna(df['Graduation rate'].median())
df['% new stud. from top 10%'] = df['% new stud. from top 10%'].fillna(df['% new stud. from top 10%'].median())
df['% new stud. from top 25%'] = df['% new stud. from top 25%'].fillna(df['% new stud. from top 25%'].median())
df['add. fees'] = df['add. fees'].fillna(df['add. fees'].median())
df['estim. personal $'] = df['estim. personal $'].fillna(df['estim. personal $'].median())
```

```python
# Checking missing values
df.isna().sum()
```

```python
# Renaming columns for better undertsanding
df = df.rename(columns={
    'Public (1)/ Private (2)': 'type',
    '# appli. rec\'d': 'applications_received',
    '# appl. accepted': 'applications_accepted',
    '# new stud. enrolled': 'students_enrolled',
    '% new stud. from top 10%': 'pct_top10',
    '% new stud. from top 25%': 'pct_top25',
    '# FT undergrad': 'fulltime_undergrads',
    '# PT undergrad': 'parttime_undergrads',
    'in-state tuition': 'tuition_in',
    'out-of-state tuition': 'tuition_out',
    'add. fees': 'fees',
    'estim. book costs': 'book_cost',
    'estim. personal $': 'personal_cost',
    '% fac. w/PHD': 'faculty_phd_pct',
    'stud./fac. ratio': 'student_faculty_ratio',
    'Graduation rate': 'grad_rate',
})
```

## Outlier Detection

```python
sns.boxplot(x=df['tuition_out'])
```

- Some colleges charge way more for out-of-state tuition than others—and a few go completely over the top. I used a boxplot to check the range and found that some schools were charging way above the rest, which made the overall data look uneven. 

- To fix that, I decided to remove the top 1% of those super expensive schools. This way, we can focus on the majority of schools and get a clearer picture of what's typical, without those extreme cases messing things up. Because honestly, a $50,000 tuition is wild—and not very helpful when we’re trying to find general trends.

```python
df = df[df['tuition_out'] < df['tuition_out'].quantile(0.99)]

df.head()
```

## Feature Engineering

```python
# Introducing few new columns for further analysis
df['acceptance_rate'] = df['applications_accepted'] / df['applications_received']
df['enrollment_yield'] = df['students_enrolled'] / df['applications_accepted']
df['total_undergrads'] = df['fulltime_undergrads'] + df['parttime_undergrads']
df['avg_total_cost'] = df[['tuition_in', 'fees', 'book_cost', 'personal_cost']].sum(axis=1)
```

```python
df.head()
```

## Data Visualizations

### Distribution of Graduation Rates

**Why:** To understand how well students graduate across institutions.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['grad_rate'], 
             bins=20, 
             color='skyblue')

plt.title("Distribution of Graduation Rates")
plt.xlabel("Graduation Rate (%)")
plt.ylabel("Number of Schools")
```

**Insight:** Most schools have graduation rates between 40–80%, with some rare overachievers and underperformers.

### Graduation Rate by Institution Type
**Why:** To compare the graduation outcomes across public (1) and private (2) institutions.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, 
            x='type', 
            y='grad_rate', 
            palette=['green', 'yellow'])

plt.title("Graduation Rate by Institution Type")
plt.xlabel("Institution Type (1 = Public, 2 = Private)")
plt.ylabel("Graduation Rate (%)")
```

**Insight:** Private institutions often show higher median graduation rates—possibly due to smaller class sizes or more resources.

### Correlation Heatmap 
**Why:** To identify the relationships between key numerical variables.

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(numeric_only=True), 
            annot=True, 
            cmap='coolwarm', 
            center=0)
plt.title("Correlation Matrix: Higher Education Metrics")
```

### Distribution of Tuition (Out-of-State)
**Why:** To visualize the tuition cost range after outlier filtering.

```python
plt.figure(figsize=(8, 5))
sns.histplot(data = df,
             x= 'tuition_out', 
             bins=30, 
             color='orange')
plt.title("Distribution of Out-of-State Tuition")
plt.xlabel("Tuition ($)")
plt.ylabel("Number of Schools")
```

**Insight:** The majority of schools fall under $20,000, which is more affordable than Ivy-like extremes.

### Faculty PhD Percentage vs Graduation Rate
**Why:** To investigate if the faculty qualifications impact student success.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, 
                x='faculty_phd_pct', 
                y='grad_rate',
                color = 'purple',
                s = 50, 
                edgecolor = 'black',
                alpha = 0.5)

plt.title("Faculty PhD % vs Graduation Rate")
plt.xlabel("Faculty with PhD (%)")
plt.ylabel("Graduation Rate (%)")
```

**Insight:** Schools with a higher percentage of PhD faculty generally show better graduation rates—possibly due to better instruction quality.

### Full-Time vs Part-Time Enrollment
**Why:** To compare the distribution of full-time and part-time undergraduates.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['fulltime_undergrads'], 
             bins=30, 
             color='green', 
             label='Full-Time', 
             kde=True, 
             binrange = (0, 25000)
            )
sns.histplot(df['parttime_undergrads'], 
             bins=30, 
             color='red', 
             label='Part-Time', 
             kde = True,
             binrange = (0, 25000)
            )

plt.title("Full-Time vs Part-Time Undergraduate Distribution")
plt.xlabel("Number of Students")
plt.ylabel("Frequency")
plt.legend()
```

**Insight:** Most schools have a much larger population of full-time students, but a few rely heavily on part-time enrollments.

### Student-Faculty Ratio vs Graduation Rate
**Why:** Test if smaller classes lead to better outcomes.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, 
                x='student_faculty_ratio', 
                y='grad_rate', 
                hue='type', 
                palette=['pink', 'yellow'],
                s = 50, 
                edgecolor = 'black',
                alpha = 0.5)

plt.title("Student-Faculty Ratio vs Graduation Rate")
plt.xlabel("Student-Faculty Ratio")
plt.ylabel("Graduation Rate (%)")
```

**Insight:** Lower student-faculty ratios may be associated with higher graduation rates, especially among private schools.

### Graduation Rate vs Average Total Cost
**Why:** To check if more expensive schools graduate students more successfully.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, 
                x='avg_total_cost', 
                y='grad_rate', 
                hue='type', 
                palette='muted',
                s = 50, 
                edgecolor = 'black',
                alpha = 0.5)
plt.title("Graduation Rate vs Average Total Cost")
plt.xlabel("Average Total Cost ($)")
plt.ylabel("Graduation Rate (%)")
```

**Insight:** While expensive schools (often private) may offer better graduation outcomes, the relationship isn’t strongly linear—value for money varies widely.

### Enrollment Yield vs Acceptance Rate
**Why:** Yield tells us how many accepted students actually enroll. Plotting against acceptance rate shows prestige perception.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, 
                x='acceptance_rate', 
                y='enrollment_yield', 
                hue='type',
                s = 50, 
                edgecolor = 'black',
                alpha = 0.5)
plt.title("Acceptance Rate vs Student-Faculty Ratio")
plt.xlabel("Student-Faculty Ratio")
plt.ylabel("Acceptance Rate (%)")
```

**Insight:** Some schools with low acceptance rates have high yields—this often signals high demand (think: top-tier institutions). A strong inverse pattern may emerge.

## Statistical Analysis & Key Insights

1. **Graduation Rate Patterns**
    - The graduation rate is normally distributed, with most schools clustered between 40% to 80%.

    - Private institutions (Type 2) show higher median graduation rates than public ones.

    - Students at private institutions are statistically more likely to graduate, possibly due to smaller class sizes or better resources.

2. **Faculty Credentials**
    - There’s a positive but scattered correlation between % of faculty with PhDs and graduation rates.

    - This suggests a general trend where institutions with more highly educated faculty may offer better academic support, but it's not a perfect predictor.

3. **Student-Faculty Ratio**
    - A lower student-faculty ratio is weakly associated with higher graduation rates, as seen in the scatterplot and heatmap.

    - Insight: More faculty attention may contribute to student success, but the relationship is not very strong.

4. **Cost vs Graduation**
    - Higher average total costs are positively correlated with higher graduation rates.

    - This may indicate that more expensive institutions (often private) offer better support or attract more prepared students — but beware of assuming causation.

5. **Out-of-State Tuition**
    - Tuition costs vary widely, with most institutions charging between $5,000 and $15,000 for out-of-state students.

    - The long right tail suggests a small number of elite or private schools charging significantly more.

    - Outliers above the 99th percentile were removed to normalize the dataset for better trend analysis.

6. **Full-Time vs Part-Time Enrollment**
    - Most students are full-time, but part-time enrollment still plays a substantial role.

    - Schools with a high part-time population might offer flexibility but could see lower graduation rates due to extended timelines.

7. **Acceptance Rate Trends**
   - Schools with lower acceptance rates tend to have lower student-faculty ratios, aligning with the idea of more selective, resource-rich institutions offering a more personalized experience.
